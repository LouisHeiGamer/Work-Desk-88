```python
# Question 1

# Write a program that reads a score between 1 and 100 for a quiz from the user, assigns a letter grade (according to the following table) using if statement, and print the grade.
# A: 80 - 100
# B: 70 - 79
# C: 60 - 69
# D: 50 - 59
# E: 40 - 49
# F: < 40
```

```python
# input zone
score = int(input("Please enter your score: "))

# main part
if score < 1:
  print("Error input! Please enter a score between 1 and 100.")
elif score < 40:
  print("F")
elif score < 50:
  print("E")
elif score < 60:
  print("D")
elif score < 70:
  print("C")
elif score < 80:
  print("B")
elif score <= 100:
  print("A")
else:
  print("Error input! Please enter a score between 1 and 100.")
```

```python
# Question 2

# At an academic institution, the tuition for a full-time student in 2022 is $48,000 per year. It has been announced that the tuition will increase by 6 percent each year for the next 5 years. Write a program with a loop that displays the projected tuition amount for the next 5 years.
```

```python
# variable
tuition = 48000

# main part
for i in range(1, 6):
  tuition = tuition * (1 + 0.06)
  print(f"The projected tuition amount for the next {i} year(s) is {int(tuition)}.")
```

```python
# Question 3

# Write a program that reads an integer and displays, using asterisks(*), the following pattern of the entered side. For example, if the input is 5, the program should displays:
# *
# **
# ***
# ****
# *****
# ****
# ***
# **
# *
```

```python
# input zone
user_input = int(input("Please enter an integer: "))

# main part
for i in range(user_input):
  print(f"{'*' * (i+1)}")
for i in range(user_input - 1, 0, -1):
  print(f"{'*' * (i)}")
```

```python
# Question 4

# Write a Python function, bmi, that reads two float numbers, weight and height, and returns the BMI value computed from the weight and height. For example, you cam bmi(65, 1.7), the program should display:
#   Your BMI is 22.49
```

```python
# input zone
user_weight = float(input("Please enter your weight in KG: "))
user_height = float(input("Please enter your height in m: "))

# function
def bmi (weight, height):

  # Boundary Condition Check
  if weight <= 0 or height <= 0:
    raise ValueError("Please enter a positive value!")

  bmi = weight / height**2
  return bmi

# main part
user_bmi = bmi(user_weight, user_height)
print(f"\tYour BMI is {user_bmi:.2f}")
```
