# 戴克斯特拉演算法 (Dijkstra's algorithm)

# 一致代價搜索 (Uniform-Cost Search)

They are the same except "Uniform-Cost Search" stop when the goal node is removed from priority queue.

```python
import heapq    # Priority queue implementation

graph = {
    "S": {"A": 2, "C": 5},
    "A": {"G": 1},
    "B": {"A": 4},
    "C": {"A": 5, "G": 6, "D": 2},
    "D": {"B": 4, "E": 3},
    "E": {"B": 6, "G": 3},
    "G": {"D": 7}
}

def dijkstra(graph, start):
    """Finds the shortest path from 'start' to ALL nodes."""
    pq = [(0, start)]   # (cumulative_cost, current_node)
    distances = {start: 0}

    while pq:
        cost, node = heapq.heappop(pq)
        print(cost)
        print(node)
        graph.get(node)
        for k, v in graph.get(node).items():
            print(k, v)
            d = distances.get(k, 1000)
            #print(d)
            new_cost = cost + v
            print(new_cost)
            if new_cost < d:
                distances[k] = new_cost
                heapq.heappush(pq, (new_cost, k))

    return distances

print("Dijkstra (All node from A):", dijkstra(graph, "S"))
```
