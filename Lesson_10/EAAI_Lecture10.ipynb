# 一致代價搜索 (Uniform-Cost Search)

They are the same except "Uniform-Cost Search" stop when the goal node is removed from priority queue.

```python
# Uniform-Cost Search (Dijkstra's Algorithm) Implementation in Python

import heapq    # Priority queue implementation

# Cost graph
graph = {
    "S": {"A": 2, "C": 5},
    "A": {"G": 1},
    "B": {"A": 4},
    "C": {"A": 5, "G": 6, "D": 2},
    "D": {"B": 4, "E": 3},
    "E": {"B": 6, "G": 3},
    "G": {"D": 7}
}

def dijkstra(graph, start, goal):
    """Finds the shortest path from 'start' to 'goal'."""
    pq = [(0, start, [start])]   # (cumulative_cost, current_node, path)
    distances = {start: 0}

    while pq:
        cost, node, path = heapq.heappop(pq)
        #print(cost)
        #print(node)
        
        if node == goal:
            return path

        graph.get(node)
        for k, v in graph.get(node).items():
            #print(k, v)
            d = distances.get(k, 1000)
            #print(d)
            new_cost = cost + v
            #print(new_cost)

            if new_cost < d:
                distances[k] = new_cost
                heapq.heappush(pq, (new_cost, k, path + [k]))

    return None

print("Dijkstra (S to G):", dijkstra(graph, "S", "G"))  # Change the start and goal nodes to see different paths
```
