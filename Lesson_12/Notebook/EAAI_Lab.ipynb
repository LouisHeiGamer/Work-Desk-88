```python
## Q.1

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': [],
}

visited = []    # List to keep track of visited nodes.
queue = []      # Initialize a queue

def bfs(visited, graph, node):
    visited.append(node)
    print("visit:" + node, end = " ")
    queue.append(node)
    print("Queue:", queue)

    while queue:
        s = queue.pop(0)
        print("Dequeue:" + s, end = " ")

    for neighbour in graph[s]:
        if neighbour not in visited:
            visited.append(neighbour)
            queue.append(neighbour)
            print("Enqueue:" + neighbour, end = " ")
            
# Driver Code
bfs(visited, graph, 'A')
```
