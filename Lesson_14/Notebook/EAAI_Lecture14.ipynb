# Constraint Satisfaction Problem

```python
## Simple CSPs with Python

from constraint import *  # pip install python-constraint
problem = Problem()
problem.addVariable('a', range(10))
problem.addVariable('b', range(10))
problem.addConstraint(lambda a, b: a*2 == b)
solutions = problem.getSolutions()
print(solutions)
```

```python
## Magic Square with CSP (n = 3)
#import string

n = 3
problem = Problem()
problem.addVariables(range(0, 9), range(1, 10))
problem.addConstraint(AllDifferentConstraint(), range(0, 9))
problem.addConstraint(ExactSumConstraint(15, [0, 4, 8]))  # Left Diagonal
problem.addConstraint(ExactSumConstraint(15), [2, 4, 6])  # Right Diagonal

for row in range(3):
    problem.addConstraint(ExactSumConstraint(15), [row*3+i for i in range(3)])

for col in range(3):
    problem.addConstraint(ExactSumConstraint(15), [col+3*i for i in range(3)])

solutions = problem.getSolutions()
print(solutions)
print('Solution found : %i' % len(solutions))
```

```python
## N-Queens Problem

n = 5
problem = Problem()
cols = range(1, n+1)
rows = range(1, n+1)
problem.addVariables(cols, rows)

for col1 in cols:
    for col2 in cols:
        if col1 < col2:
            problem.addConstraint(lambda row1, row2, col1=col1, col2=col2:
                                  abs(row1-row2) != abs(col1-col2) and
                                  row1 != row2, (col1, col2))

solutions = problem.getSolutions()
print(solutions)
print('Solutions found : %i' % len(solutions))
```

```python
## SEND + MORE = MONEY
from constraint import *

letters = ['D', 'E', 'M', 'N', 'O', 'R', 'S', 'Y']
digits = range(10)

def sum_constraint(D, E, M, N, O, R, S, Y):
    send = S * 1000 + E * 100 + N * 10 + D
    more = M * 1000 + O * 100 + R * 10 + E
    money = M * 10000 + O * 1000 + N * 100 + E * 10 + Y
    return send + more == money

problem = Problem()
for l in letters:
    problem.addVariable(l, digits)

problem.addConstraint(AllDifferentConstraint(), letters)
problem.addConstraint(sum_constraint, letters)
# problem.addConstraint(lambda x: x != 0, ['m'])

solutions = problem.getSolutions()
for s in solutions:
    print(s)
```
