### Ex 1 - Magic Square

```python
# Magic Square with N Using CSP

n = 3

csum = n*(n*n+1)/2
ldiag = [(n+1)*i for i in range(n)]  # Left diagonal
rdiag = [(n-1)*(i+1) for i in range(n)]  # Right diagonal

problem = Problem()
problem.addVariables(range(0, n*n), range(1, n*n+1))
problem.addConstraint(AllDifferentConstraint(), range(0, n*n))
problem.addConstraint(ExactSumConstraint(csum), ldiag)
problem.addConstraint(ExactSumConstraint(csum), rdiag)

for row in range(n):
    problem.addConstraint(ExactSumConstraint(csum),
                          [row*n+i for i in range(n)])

for col in range(n):
    problem.addConstraint(ExactSumConstraint(csum),
                          [col+n*i for i in range(n)])

solutions = problem.getSolution()
print(solutions)
```

### Ex 2 - N-Rook Problem

```python
n = 5
cols = range(1, n+1) # Variables
rows = range(1, n+1) # Domains

problem = Problem()
problem.addVariables(cols, rows)

# Goal: Assign domain values in rows (1~n) to variables in cols (1~n)

for col1 in cols:
    for col2 in cols:
        if col1 < col2:
            problem.addConstraint(lambda row1, row2, col1=col1, col2=col2:
                row1 != row2, (col1, col2))

solutions = problem.getSolutions()
for sol in solutions:
    print(sol)
print('Solutions found : %i' % len(solutions))
```

### Ex 3 - Factorial

```python
n = var()

@match(0)
def factl(n):
    return 1

@match(n)
def factl(n):
    return n * factl(n-1)
```

### Ex 4 - Relations in Logic Programming

```python
from kanren import Relation, fact, run, var
human = Relation()
fact(human, 'bob')
run(0, True, human('bob'))
```

```python
# Given Code
parent = Relation()
facts(parent, ('Tom', 'Bobby'),
              ('Mary', 'Bobby'),
              ('Tom', 'Amy'),
              ('Mary', 'Amy'))
```

```python
# Ex 4(a)
male = Relation()
fact(male, 'Tom')
fact(male, 'Bobby')

female = Relation()
fact(female, 'Mary')
fact(female, 'Amy')
```

```python
# Ex 4(b)
def father(x, y):
    return lall(parent(x, y), male(x))
```

```python
# Ex 4(c)
def mother(x, y):
    return lall(parent(x, y), female(x))
```

```python
# Ex 4(d)
def brother(x, y):
    z = var()
    return lall(parent(z, x), male(x), neq(x, y), parent(z, y))
```

```python
# Ex 4(e)
def sister(x, y):
    z = var()
    return lall(parent(z, x), female(x), neq(x, y), parent(z, y))
```

#### Queries for the relations

```python
x = var()
run(0, x, brother('Bobby', x))
```

```python
x = var()
run(0, x, sister('Amy', x))
```

```python
x = var()
run(0, x, sister(x, 'Bobby'))
```
