### Ex 1

```python
# Program to compute the bill ammount of meal at a restaurant
charge = int(input('Enter the charge of food: '))
tip = charge * 0.1
tax = charge * 0.05
bill_amt = charge + tip + tax
print('Charge amount: $', format(charge, ',.2f'), sep='')
print('Tip amount: $', format(tip, ',.2f'), sep='')
print('Tax amount: $', format(tax, ',.2f'), sep='')
print('Bill amount: $', format(bill_amt, ',.2f'), sep='')
```

### Ex 2

```python
# Program to compute the ingredients of making cookie
n = int(input('How many cookies you want to make? '))
sugar = n * 240 / 48
butter = n * 200 / 48
flour = n * 560 / 48
print('The amount of sugar needed: ', format(sugar, '.0f'), ' cc', sep='')
print('The amount of butter needed: ', format(butter, '.0f'), ' cc', sep='')
print('The amount of flour needed: ', format(flour, '.0f'), ' cc', sep='')
```

### Ex 3

```python
# Program to reads a score from the user, assigns a letter grade, and print the grade.
score = int(input("Enter score between 1 and 100: "))
if score < 40:
    grade = 'F'
elif score < 50:
    grade = 'E'
elif score < 60:
    grade = 'D'
elif score < 70:
    grade = 'C'
elif score < 80:
    grade = 'B'
else:
    grade = 'A'

print("Your grade is %s" %(grade))
```

### Ex 5

```python
ch = '*'
n = int(input('Enter the size of the pattern: '))
```

```python
# Program to print a specific pattern of asterisks

# Display upper part
for i in range(n-1):
    for c in range(i+1):
        print(ch, end='')
    print()

# Display middle line
for m in range(n):
    print(ch, end='')
print()

# Display lower part
for j in range(n-1):
    for c in range(n-j-1):
        print(ch, end='')
    print()
```

```python
# Program to print a specific pattern of asterisks (Method 2)

# Display upper part
for i in range(1, n+1):
    print(ch*i)

# Display lower part
for i in range(n-1, 0, -1):
    print(ch*i)
```

```python
# Program to print a specific pattern of asterisks (Method 3)
for i in list(range(1, n+1)) + list(reversed(range(1, n))):
    print(ch * i)
```
