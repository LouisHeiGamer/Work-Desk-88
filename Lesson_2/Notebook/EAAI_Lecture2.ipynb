## 單元二：數據類型（Data Types）與物件觀念

### 1. 物件（Objects）與動態型別（Dynamic Typing）

- 在 Python 中，**一切皆為物件（Objects）**。每個物件包含三個基本屬性：
    - Identity（身份/記憶體位址）：物件創建後即固定，可用 id() 檢視。
    - Type（類型）：決定物件能進行哪些操作，可用 type() 檢視。
    - Value（值）：物件所儲存的資料內容。

- 動態型別（Dynamic Typing）：
    - 不需要事先宣告變數類型，在賦值（Assignment）時自動判定。

### 2. 賦值（Assignmnent）與比較算子 (= vs == vs is)
- 賦值算子 = ：將右邊的值指派給左邊的變數。
- 相等算子 ==：比較兩者的「數值/內容」是否相等。
- 身份算子 is：比較兩者是否為「同一個記憶體物件（Identity）」。

```python
## Assignment vs. Comparison Operators

a = 5
b = 5.0
c = 5

print(a == b)   # True （值都是 5）
print(a is b)   # False（a 是整數 int，b 是浮點數 float，屬於不同物件）
print(a is c)   # True （a 和 c 都是整數 int，且值都是 5，屬於同一個物件）


# Extra Thinking

a = a - 1

print(a is c)   # False（a 的值變成 4，c 的值仍然是 5，屬於不同物件）
```

### 3.可變性（Mutability）

- 不可變（Immutable）：物件創建後無法直接修改其內容（如修改會創建新物件）。包含：Numbers（數字）、Strings（字串）、Tuples（元組）。
- 可變（Mutable）：物件創建後可以直接修改內部元素。包含：Lists（列表）、Dictionaries（字典）、Sets（集合）。

## 單元三：數字、字串與輸入輸出

### 1. 數字類型（Numbers）與算術運算子（Arithmetic operator）

- 數字類型（Numbers）
    - Python 主要有整數（int）與浮點數（float）。

- 算術運算子（Arithmetic operator）
    - 運算子 / ：一般除法（結果永遠為浮點數）
    - 運算子 //：整除/地板除法（Floor Division）
    - 運算子 % ：取餘數（Modulus）
    - 運算子 **：次方（Power）

```python
## Arithmetic Operators


# Division /
print(3 / 2)   # 輸出: 1.5（除法運算，結果為浮點數）

# Floor Division //
print(3 // 2)  # 輸出: 1（整除運算，結果為整數）

# Modulus %
print(3 % 2)   # 輸出: 1（取餘數運算，結果為 1）

# Exponentiation **
print(3 ** 2)  # 9（次方運算，3 的平方為 9）
```

### 2. 字串（String）與索引切片（Indexing & Slicing）

字串是字元的**不可變序列**。索引從 0 開始。

切片語法：string[start:stop:step]（包含 start，不包含 stop）。

```python
## String and Indexing & Slicing

t = "Big Data Analytics"

print(t[0])     # 'B'（索引 0 的字元，表示第一個字元）
print(t[-1])    # 's'（索引 -1 的字元，表示最後一個字元）
print(t[:5])    # 'Big D'（切片，從索引 0 到索引 4 的字元）
print(t[4:])    # 'Data Analytics'（切片，從索引 4 到最後的字元）
print(t[4:8])   # 'Data'（切片，從索引 4 到索引 7 的字元）
print(t[::2])   # 'BgDt nltc'（切片，從索引 0 開始，每隔一個字元取一個字元）
print(t[::-1])  # 'scitylanA ataD giB'（切片，將字串反轉）

# Extra Thinking
print(t[::])    # Big Data Analytics（切片，從索引 0 開始，到最後的字元，步長為 1，結果為原字串）
print(t[::1])   # Big Data Analytics（切片，從索引 0 開始，到最後的字元，步長為 1，結果為原字串）
print(t[::-1][::-1])  # Big Data Analytics（切片，將字串反轉兩次，結果為原字串）
print(t[4:8][::-1])  # ataD（切片，先取索引 4 到索引 7 的字元，再將結果反轉）
print(t[4:8][::-1][::-1])  # Data（切片，先取索引 4 到索引 7 的字元，再將結果反轉兩次，結果為原字串）
```

### 3. 印出格式化（Print Formatting）

使用 % 算子進行格式化輸出：

- %s：字串
- %d：整數
- %f：浮點數（語法：%[總寬度].[小數點位數]f）

```python
## Print Formatting
pi = 3.14159
print("圓周率: %1.2f" % pi)  # 輸出: 圓周率: 3.14（格式化輸出，保留兩位小數）
print("整數排版: %5d" % 123)  # 輸出: 整數排版:   123（格式化輸出，總寬度為 5，右對齊，前面補兩個空格）


# Extra Thinking
print("圓周率: {:.2f}".format(pi))  # 輸出: 圓周率: 3.14（格式化輸出，保留兩位小數）
print(f"圓周率: {pi:.2f}")  # 輸出: 圓周率: 3.14（格式化輸出，保留兩位小數，使用 f-string）
print(f"圓周率: {pi:10.2f}")  # 輸出: 圓周率:       3.14（格式化輸出，保留兩位小數，總寬度為 10，右對齊）
print(f"圓周率: {pi:<10.2f}")  # 輸出: 圓周率: 3.14      （格式化輸出，保留兩位小數，總寬度為 10，左對齊）
print(f"圓周率: {pi:^10.2f}")  # 輸出: 圓周率:    3.14    （格式化輸出，保留兩位小數，總寬度為 10，置中對齊）
print(f"圓周率: {pi:*>10.2f}")  # 輸出: 圓周率: ******3.14（格式化輸出，保留兩位小數，總寬度為 10，右對齊，使用 * 填充）
```

## 4. 讀取使用者輸入與型別轉換（Casting）

input() 函式讀入的數據類型一律為字串（str），若需要數學計算必須先轉型。

```python
## Casting

val = input("請輸入一個數字: ")  # 假設輸入 123
print(val * 2)  # 輸出: '123123'（輸入的值是字串，乘以 2，結果是字串重複兩次）
print(int(val) * 2)  # 輸出: 246（將輸入的字串轉換為整數，乘以 2，結果是整數）
```
