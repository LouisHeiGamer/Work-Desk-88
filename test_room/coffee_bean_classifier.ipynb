## 綜合測驗：咖啡豆品種分類器 (Coffee Bean Classifier)

這是一個標準的基礎機器學習微型專案。你將運用剛剛學過的所有技巧，替一家咖啡烘焙坊開發一個能自動區分「阿拉比卡 (Arabica)」與「羅布斯塔 (Robusta)」咖啡豆的模型。

### 建立這份帶有「人工瑕疵」的模擬資料集：

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

# 阿拉比卡豆：通常顆粒較大，咖啡因含量較低
arabica = pd.DataFrame({
    "diameter": rng.normal(12.0, 1.5, 50).round(1),
    "caffeine": rng.normal(1.2, 0.3, 50).round(2),
    "bean_type": "arabica"
})

# 羅布斯塔豆：通常顆粒較小，咖啡因含量較高
robusta = pd.DataFrame({
    "diameter": rng.normal(8.0, 1.2, 50).round(1),
    "caffeine": rng.normal(2.2, 0.4, 50).round(2),
    "bean_type": "robusta"
})

coffee = pd.concat([arabica, robusta], ignore_index=True)

# 模擬真實世界的髒數據 (Dirty Data)
coffee.loc[[12, 34, 67, 88], "bean_type"] = ["Arabica", "arabica ", "ROBUSTA", " Robusta"]
coffee.loc[[5, 25, 60, 95], "diameter"] = np.nan
coffee = pd.concat([coffee, coffee.loc[[10, 45, 80]]], ignore_index=True)

# 將這份資料洗牌，增加一點真實感
coffee = coffee.sample(frac=1, random_state=42).reset_index(drop=True)
```

2. 資料視覺化 (Data Visualization)

3. 資料拆分 (Train/Test Split)
    - 將資料拆分為特徵矩陣 X 與目標標籤 y。

4. 特徵預處理 (Feature Preprocessing)

5. 模型訓練與評估 (Model Training & Evaluation)
    - 建立一個 $k=3$ 的 KNN 分類器。

6. 超參數優化 (Hyperparameter Tuning)
    - 寫一個簡單的迴圈，測試 $k=1, 3, 5, 7, 9$ 時的測試集準確率（Accuracy），並找出表現最好的是哪一個 $k$ 值。
